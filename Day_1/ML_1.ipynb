```python
import math

words = ["learn", "build", "sleep", "banana"]
scores = [3.0, 2.0, 0.5, -1.0]

def get_probabilities(scores, temperature):
  scaled = []
  for score in scores:
    scaled.append(math.exp(score/temperature))
  total = sum(scaled)
  probabilities = []
  for value in scaled:
    probabilities.append(value/total)
  return probabilities

def apply_top_p(words, probabilities, top_p):
  ranked = sorted(zip(probabilities, words), reverse = True)
  kept = []
  running_total = 0
  for probability, word in ranked:
    kept.append(word)
    running_total += probability
    if running_total >= top_p:
      break
  return kept

for temperature in [0.5, 1.5]:
  probs = get_probabilities(scores, temperature)
  print(f"Temperature: {temperature}: ")
  for word, probability in zip(words, probs):
    print(f"{word:<7} {probability:.2%}")

probs = get_probabilities(scores, 1.0)
print("Top-P: 0.9 keeps:", apply_top_p(words, probs, 0.9))
print("Top-P: 0.5 keeps:", apply_top_p(words, probs, 0.5))
```

```python
import os
from groq import Groq

MODEL_NAME = "qwen/qwen3.8-27b"

def get_groq_key():
    try:
        from google.colab import userdata
        return userdata.get("GROQ_API_KEY")
    except Exception:
        return os.environ.get("GROQ_API_KEY")

api_key = get_groq_key()

if not api_key:
    print("GROQ_API_KEY not found. Colab: click the key icon, add GROQ_API_KEY, and enable Notebook access.")
else:
    client = Groq(api_key=api_key)
    response = client.chat.completions.create(
        model=MODEL_NAME,
        messages=[
            {"role": "system", "content": "You are a clear teacher for engineering students."},
            {"role": "user", "content": "Explain what an LLM is in two sentences?"}
        ],
        temperature=0.3,
        max_tokens=200
        )

    print(response.choices[0].message.content)
    print("Tokens used:", response.usage.total_tokens)
```
